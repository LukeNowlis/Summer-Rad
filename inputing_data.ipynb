def clip_photon(photon, start_time, window_size, delay_steps=20):

This is goin to be the most complicated step in the process
it takes all the photon data in the file we chose and clips it to be just the data points that we want in the profile 

To do this we give it the starting index of the beam profile clip_photon(photon,#This is the starting index,5)

To find the starting index I reconmend vizulaling the code using the medthod I talked about earlier and adjusting the window until it looks like the beam profile you are intrested in, the code should print out what index you used to adjust the start of the window and you can use this value in you clip_photon call 

IMPORTANT
when choosing your starting index make sure you chose it to be higher than or equal to the point you want the profile to end at. The profile will clasify the profile as all the data until the point where the photon values drop back to the value of the starting index. 

This function works by starting at the start index and checking when the photon value gets back to that level as it would in an expected beam profile. It does this by using a rolling window and taking the average in that window and the number of values included in that window is the input window_size  clip_photon(photon,90,#This one here) you can adujst this based on the number of points in your beam profile. There should be much need to go higher than 5 but for a profile with few data points it should be dropped to 3 or even 1. 

Same goes for the delay_steps=20 input. This skips the first 20 indexs after the starting index becasue of normal variations but if you have less than 20 data points in your profile this number must be adjusted.

def interpolate_position_data(inline, crossline, depth, target_length, type='line'):


This function takes the position data we got from the file and makes it the same length as the number of photon values we have so that we are able to graph it. 

There are two ways that this function can do this. 

Option 1- type ='line'
This takes the first and last position values and the length of the photon list of photon values that we want and evenly spaces out the points between the start and end so that the length of the position values equals the length of the photon values. 

use this option if you are using continous scans or your position data seems to not be quite right. This is the simpilist form of interpolating the data and will likely do the best job although it may not be perfect. 

Option 2- type= 'read'
This takes into account every value from the position list and interpolates what the values in between these should be so that the length matches the photon length. The problem with this is that if the position data isn't perfect this option may just be highlighting those imperfections more

def beam_profile(datanum, photon, position, stime, beam_size, profile_type='inline', fit_type='super_gaussian', avg_group=10):

This is mostly just imputing the values we already have got 

make sure for the position input you are using the interpolated value or it will not run. You also should use the position data the you need i.e. for an inline profile use in_1_interp but for a cross line it should be cr_1_interp. 

stime is just a string input that tells you the time at witch the scan started withc can be hwlpful for keeping track of what beam profile you are doing

beam_size is the dimentions of the beam you are scanning and it will be displayed on the graph, input this in mm 

Profile_type in just in_line or cross_line 

I would leave the fit_type as super_gaussian as it geneally give a better beam profile but you can experirament with changing it to gaussian if you so choose
